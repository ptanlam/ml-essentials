```python
import pandas as pd
import matplotlib.pyplot as plt

import seaborn as sns
```

```python
data = pd.read_csv('dataset/diamonds.csv', index_col=0)

data.head()
```

```python
data.shape
```

```python
correlation = data.corr()

correlation
```

```python
plt.subplots(figsize=(10,8))
sns.heatmap(correlation, annot=True)

plt.show()
```
