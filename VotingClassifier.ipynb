```python
import pandas as pd
```

```python
diabetes_data = pd.read_csv('processed_dataset/PimaIndians_processed.csv')

diabetes_data.head(10)
```

```python
X = diabetes_data.drop('test', axis=1)
Y = diabetes_data['test']
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2)
```

```python
x_train.shape, y_train.shape

x_test.shape, y_test.shape
```

```python
from sklearn.ensemble import VotingClassifier

from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
```

```python
log_clf = LogisticRegression(C=1, solver='liblinear')

svc_clf = SVC(C=1, kernel='linear', gamma='auto')

naive_clf = GaussianNB()
```

```python
voting_clf_hard = VotingClassifier(estimators=[('lr', log_clf), ('svc', svc_clf), ('naive', naive_clf)], voting='hard')
```

```python
voting_clf_hard.fit(x_train, y_train)
```

```python
y_pred = voting_clf_hard.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)
```

```python
for clf_hard in (log_clf, svc_clf, naive_clf, voting_clf_hard):
        clf_hard.fit(x_train, y_train)
        y_predict = clf_hard.predict(x_test)
        
        print(clf_hard.__class__.__name__, accuracy_score(y_test, y_predict))
```

```python
svc_clf_soft = SVC(C=1, kernel='linear', gamma='auto', probability=True)
```

```python
voting_clf_soft = VotingClassifier(estimators=[('lr', log_clf), ('svc', svc_clf_soft), ('naive', naive_clf)], 
                                   voting='soft', weights=[0.25,0.5,0.25])
```

```python
for clf_soft in (log_clf, svc_clf_soft, naive_clf, voting_clf_soft):
        clf_soft.fit(x_train, y_train)
        y_predict = clf_soft.predict(x_test)
        
        print(clf_soft.__class__.__name__, accuracy_score(y_test, y_predict))
```
